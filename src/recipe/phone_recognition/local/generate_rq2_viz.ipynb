```python
import pandas as pd

path = "/work/nvme/bbjs/sbharadwaj/powsm/PhoneBench/exp/rq2/phone_inventory_metrics.csv"
df = pd.read_csv(path)
df
```

```python
import matplotlib.pyplot as plt

df = df.sort_values("f1")
y = range(len(df))
plt.figure(figsize=(6, 3))
plt.hlines(y, df["precision"], df["recall"], alpha=0.6)
plt.scatter(df["precision"], y, label="Precision")
plt.scatter(df["recall"], y, label="Recall")

plt.yticks(y, df["ModelString"])
plt.xlabel("Score")
plt.xlim(0, 1)
plt.legend()
plt.tight_layout()
plt.show()
```
