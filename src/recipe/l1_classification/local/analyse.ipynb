```python
import json
import glob
import os

MODELS = [
    "ctag",
    "powsm",
    "powsm_ctc",
    "gemini",
    "xlsr53",
    "zipactc",
    "zipactc_ns",
    "lv60",
]
DATA = ["edacc"]


def get_json_path(model, data):
    path_template = "/data/user_data/sbharad2/PhoneBench/exp/runs/inf_{data}_{model}/*/transcription.json"
    paths = glob.glob(path_template.format(data=data, model=model))
    if paths:
        return max(paths, key=os.path.getmtime)
    return None


def load_transcriptions(model, data):
    path = get_json_path(model, data)
    if path is None:
        return None
    with open(path, "r") as f:
        return json.load(f)
```

```python
edacc_map = {
    0: "AFRICAN_ENGLISH",
    1: "AFROASIATIC_SEMITIC",
    2: "CARIBBEAN",
    3: "CELTIC_ENGLISH",
    4: "EAST_ASIAN",
    5: "GALLO_ROMANCE",
    6: "GERMANIC",
    7: "INNER_CIRCLE_ENGLISH",
    8: "INSULAR_SEA",
    9: "MAINLAND_SEA",
    10: "ROMANCE",
    11: "SLAVIC_BALKAN",
    12: "SOUTH_ASIAN",
}


def get_edacc_wav_file(utt_id):
    edacc_wav_files = glob.glob(
        "/data/user_data/sbharad2/PhoneBench/exp/cache/edacc/saved/*/*.wav"
    )
    for wav_file in edacc_wav_files:
        if utt_id in wav_file:
            return wav_file
    return None
```

```python
T = {}
for m in MODELS:
    t = load_transcriptions(m, "edacc")
    T[m] = t
```

```python
def display_audio(file_path):
    from IPython.display import Audio, display

    display(Audio(filename=file_path))


cnt = 0
for k in T["ctag"].keys():
    if T["ctag"][k]["passthrough"]["split"] != "test":
        continue
    if T["ctag"][k]["passthrough"]["target"] != 12:
        continue
    print("-----" * 20)
    print(k, T["ctag"][k]["passthrough"]["utt_id"])
    for m in MODELS:
        print(f"Model: {m}, Predicted: {T[m][k]['pred'][0]['processed_transcript']}")
    print("Target:", edacc_map[T["ctag"][k]["passthrough"]["target"]])
    print("WAV FILE:", get_edacc_wav_file(T["ctag"][k]["passthrough"]["utt_id"]))
    display_audio(get_edacc_wav_file(T["ctag"][k]["passthrough"]["utt_id"]))
    cnt += 1
    if cnt >= 20:
        break
```
