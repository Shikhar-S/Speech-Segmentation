```python
import collections


def split(l12speaker_duration, test_ratio):
    """
    Splits the corpus into train and test sets ensuring conversations are unique to splits.

    Args:
        l12speaker_duration: dict { l1: [(speaker_id, instances), ...] }
        test_ratio: float, proportion of speakers to put in test set.

    Returns:
        train_split: dict { l1: [(speaker_id, instances), ...] }
        test_split: dict { l1: [(speaker_id, instances), ...] }
    """

    # 1. Group speakers by their Conversation ID
    # Format: EDACC-C06-A -> Conv ID is C06
    def get_conv_id(speaker_id):
        parts = speaker_id.split("-")
        return parts[1] if len(parts) > 1 else speaker_id

    # conv_to_speakers maps conv_id -> list of (l1, speaker_id, instances)
    conv_to_speakers = collections.defaultdict(list)
    # Track which conversations belong to which L1
    l1_to_convs = collections.defaultdict(set)

    for l1, speakers in l12speaker_duration.items():
        for speaker_id, instances in speakers:
            c_id = get_conv_id(speaker_id)
            conv_to_speakers[c_id].append((l1, speaker_id, instances))
            l1_to_convs[l1].add(c_id)

    train_split = collections.defaultdict(list)
    test_split = collections.defaultdict(list)

    # Track which conversations have already been assigned (to handle cross-L1 conversations)
    assigned_convs = {}  # conv_id -> "train" or "test"

    for l1 in l12speaker_duration.keys():
        # Get all conversations associated with this L1 that haven't been assigned yet
        available_convs = [c for c in l1_to_convs[l1] if c not in assigned_convs]

        if not available_convs:
            continue

        # Calculate target number of speakers for this L1 test set
        total_speakers_in_l1 = len(l12speaker_duration[l1])
        target_test_speakers = max(1, round(total_speakers_in_l1 * test_ratio))

        # To balance instances, we sort conversations by their total instance count
        # This allows us to pick a mix of small and large conversations
        def get_conv_weight(c_id):
            return sum(s[2] for s in conv_to_speakers[c_id] if s[0] == l1)

        available_convs.sort(key=get_conv_weight, reverse=True)

        current_test_speakers = 0

        # Assign to test until we hit the speaker ratio
        for c_id in available_convs:
            # Count speakers in this conversation that belong to the current L1
            speakers_in_conv_for_l1 = [s for s in conv_to_speakers[c_id] if s[0] == l1]
            num_speakers = len(speakers_in_conv_for_l1)

            if current_test_speakers + num_speakers <= target_test_speakers:
                assigned_convs[c_id] = "test"
                current_test_speakers += num_speakers
            else:
                assigned_convs[c_id] = "train"

    # 2. Final Distribution
    # Now that all conversations are tagged, populate the dictionaries
    for c_id, split_type in assigned_convs.items():
        for l1, speaker_id, instances in conv_to_speakers[c_id]:
            if split_type == "test":
                test_split[l1].append(speaker_id)
            else:
                train_split[l1].append(speaker_id)

    return dict(train_split), dict(test_split)
```
